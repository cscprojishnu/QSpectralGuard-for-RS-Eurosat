```python
# =============================================================================
# Q-SPECTRALGUARD
# Quantum-Enhanced Spectral-Spatial Learning for EuroSAT
#
# ONE-CELL COMPLETE EXPERIMENT
#
# Dataset:
#   EuroSAT multispectral Sentinel-2
#
# Pipeline:
#   Multispectral Input
#        ↓
#   CNN Spectral-Spatial Encoder
#        ↓
#   Spectral Bottleneck
#        ↓
#   Quantum-inspired trainable feature map
#        ↓
#   Classifier
#
# Outputs:
#   ✓ Train/Validation/Test split
#   ✓ Per-epoch train loss
#   ✓ Per-epoch train accuracy
#   ✓ Per-epoch validation loss
#   ✓ Per-epoch validation accuracy
#   ✓ Best model checkpoint
#   ✓ Test accuracy
#   ✓ Precision
#   ✓ Recall
#   ✓ F1-score
#   ✓ Macro / weighted metrics
#   ✓ Classification report
#   ✓ Confusion matrix
#   ✓ Training loss graph
#   ✓ Training accuracy graph
#   ✓ Precision/Recall/F1 graph
#   ✓ ROC-AUC curve
#   ✓ Precision-Recall curve
#   ✓ Per-class F1
#   ✓ Per-class precision
#   ✓ Per-class recall
#   ✓ Sample predictions
#   ✓ All results saved to /kaggle/working/Q_SpectralGuard_Results
# =============================================================================

import os
import glob
import random
import time
import json
import copy
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)
from sklearn.preprocessing import label_binarize

from tqdm.auto import tqdm


# =============================================================================
# 1. REPRODUCIBILITY
# =============================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

torch.backends.cudnn.deterministic = False
torch.backends.cudnn.benchmark = True


# =============================================================================
# 2. DEVICE
# =============================================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 90)
print("Q-SPECTRALGUARD")
print("=" * 90)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "VRAM:",
        round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2),
        "GB"
    )

print("PyTorch:", torch.__version__)


# =============================================================================
# 3. CONFIGURATION
# =============================================================================

EPOCHS = 100
BATCH_SIZE = 64
NUM_WORKERS = min(4, os.cpu_count() or 2)

IMG_SIZE = 64
NUM_BANDS = 13
NUM_CLASSES = 10

LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4

# Number of compact spectral features entering quantum-inspired block
QUANTUM_DIM = 8

OUTPUT_DIR = "/kaggle/working/Q_SpectralGuard_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

CHECKPOINT_PATH = os.path.join(
    OUTPUT_DIR,
    "best_Q_SpectralGuard.pt"
)


# =============================================================================
# 4. AUTOMATIC DATASET DISCOVERY
# =============================================================================

print("\nSearching for EuroSAT dataset...")

possible_roots = [
    "/kaggle/input/eurosat-dataset",
    "/kaggle/input/eurosat-dataset/EuroSAT",
    "/kaggle/input/eurosat",
    "/kaggle/input/eurosat/EuroSAT",
    "/kaggle/input/eurosat-dataset/EuroSATallBands",
    "/kaggle/input/eurosat/EuroSATallBands",
]

DATASET_ROOT = None

for root in possible_roots:
    if os.path.exists(root):
        DATASET_ROOT = root
        break

# Recursive fallback
if DATASET_ROOT is None:
    candidates = glob.glob(
        "/kaggle/input/**/EuroSAT*",
        recursive=True
    )

    if len(candidates) > 0:
        DATASET_ROOT = candidates[0]

if DATASET_ROOT is None:
    raise FileNotFoundError(
        """
EuroSAT dataset was not found.

Please add the Kaggle dataset:
https://www.kaggle.com/datasets/apollo2506/eurosat-dataset

Then run this cell again.
"""
    )

print("Dataset root:", DATASET_ROOT)


# =============================================================================
# 5. FIND MULTISPECTRAL DATA
# =============================================================================

class_names = [
    "AnnualCrop",
    "Forest",
    "HerbaceousVegetation",
    "Highway",
    "Industrial",
    "Pasture",
    "PermanentCrop",
    "Residential",
    "River",
    "SeaLake"
]

class_to_idx = {
    cls: i for i, cls in enumerate(class_names)
}


# -----------------------------------------------------------------------------
# Search for GeoTIFF / TIF multispectral files
# -----------------------------------------------------------------------------

all_tifs = []

for root, dirs, files in os.walk(DATASET_ROOT):
    for file in files:
        if file.lower().endswith((".tif", ".tiff")):
            all_tifs.append(os.path.join(root, file))

print("Multispectral TIF files found:", len(all_tifs))


# =============================================================================
# 6. DATASET CLASS
# =============================================================================

class EuroSATMultispectral(Dataset):

    def __init__(self, samples):

        self.samples = samples

        self.normalize = True

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):

        path, label = self.samples[idx]

        # ---------------------------------------------------------------------
        # Read multispectral image
        # ---------------------------------------------------------------------

        try:

            import rasterio

            with rasterio.open(path) as src:
                image = src.read()

            image = image.astype(np.float32)

        except Exception:

            # fallback using tifffile
            import tifffile

            image = tifffile.imread(path).astype(np.float32)

            if image.ndim == 3:

                # Convert H,W,C → C,H,W
                if image.shape[-1] == NUM_BANDS:
                    image = np.transpose(image, (2, 0, 1))

        # ---------------------------------------------------------------------
        # Ensure 13 bands
        # ---------------------------------------------------------------------

        if image.ndim != 3:
            raise ValueError(
                f"Unexpected image shape {image.shape} for {path}"
            )

        # Sometimes loaded as H,W,C
        if image.shape[0] != NUM_BANDS and image.shape[-1] == NUM_BANDS:
            image = np.transpose(image, (2, 0, 1))

        # If fewer/more bands occur, retain first 13
        if image.shape[0] >= NUM_BANDS:
            image = image[:NUM_BANDS]

        if image.shape[0] != NUM_BANDS:
            raise ValueError(
                f"Expected {NUM_BANDS} bands, got {image.shape[0]}: {path}"
            )

        # ---------------------------------------------------------------------
        # Per-band robust normalization
        # ---------------------------------------------------------------------

        for b in range(image.shape[0]):

            band = image[b]

            low = np.percentile(band, 2)
            high = np.percentile(band, 98)

            if high > low:

                band = np.clip(
                    band,
                    low,
                    high
                )

                band = (
                    band - low
                ) / (
                    high - low + 1e-8
                )

            image[b] = band

        image = torch.tensor(
            image,
            dtype=torch.float32
        )

        # ---------------------------------------------------------------------
        # Spatial normalization
        # ---------------------------------------------------------------------

        if image.shape[-1] != IMG_SIZE or image.shape[-2] != IMG_SIZE:

            image = F.interpolate(
                image.unsqueeze(0),
                size=(IMG_SIZE, IMG_SIZE),
                mode="bilinear",
                align_corners=False
            ).squeeze(0)

        return image, torch.tensor(
            label,
            dtype=torch.long
        )


# =============================================================================
# 7. BUILD SAMPLE LIST
# =============================================================================

samples = []

if len(all_tifs) > 0:

    for path in all_tifs:

        label = None

        for cls in class_names:

            if os.sep + cls + os.sep in path:
                label = class_to_idx[cls]
                break

        if label is not None:
            samples.append((path, label))

else:

    raise RuntimeError(
        """
No multispectral TIFF files were found.

The Kaggle dataset must contain the EuroSAT multispectral/all-band files.
Please verify that you added the multispectral EuroSAT dataset.
"""
    )


print("\nTotal multispectral samples:", len(samples))


# =============================================================================
# 8. CLASS DISTRIBUTION
# =============================================================================

labels = np.array([x[1] for x in samples])

print("\nClass distribution:")

for i, cls in enumerate(class_names):
    print(
        f"{cls:25s}:",
        int(np.sum(labels == i))
    )


# =============================================================================
# 9. STRATIFIED TRAIN / VALIDATION / TEST SPLIT
# =============================================================================

indices = np.arange(len(samples))

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.30,
    random_state=SEED,
    stratify=labels
)

temp_labels = labels[temp_idx]

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_labels
)

train_samples = [samples[i] for i in train_idx]
val_samples = [samples[i] for i in val_idx]
test_samples = [samples[i] for i in test_idx]

print("\nDataset split:")
print("Train:", len(train_samples))
print("Validation:", len(val_samples))
print("Test:", len(test_samples))


# =============================================================================
# 10. DATASETS
# =============================================================================

train_dataset = EuroSATMultispectral(train_samples)
val_dataset = EuroSATMultispectral(val_samples)
test_dataset = EuroSATMultispectral(test_samples)


# =============================================================================
# 11. DATALOADERS
# =============================================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0)
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0)
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    persistent_workers=(NUM_WORKERS > 0)
)


# =============================================================================
# 12. QUANTUM-INSPIRED FEATURE MAP
# =============================================================================

class QuantumFeatureMap(nn.Module):

    """
    Differentiable quantum-inspired feature transformation.

    Structure:

        Classical feature vector
                 ↓
          Angle embedding
                 ↓
        Parameterized rotations
                 ↓
       Entanglement-like mixing
                 ↓
          Measurement-like
          nonlinear projection

    This module is intentionally simulator/GPU friendly.
    """

    def __init__(self, dim):

        super().__init__()

        self.dim = dim

        self.theta = nn.Parameter(
            torch.randn(dim) * 0.02
        )

        self.phi = nn.Parameter(
            torch.randn(dim) * 0.02
        )

        self.gamma = nn.Parameter(
            torch.randn(dim) * 0.02
        )

        self.mix = nn.Linear(
            dim,
            dim,
            bias=False
        )

    def forward(self, x):

        # Angle encoding
        q = torch.sin(
            x + self.theta
        )

        # Parameterized rotation
        q = q * torch.cos(
            self.phi
        )

        # Entanglement-inspired nearest-neighbour interaction
        rolled = torch.roll(
            q,
            shifts=1,
            dims=1
        )

        q = torch.sin(
            q +
            self.gamma *
            rolled
        )

        # Trainable mixing
        q = self.mix(q)

        # Measurement-inspired expectation
        q = torch.tanh(q)

        return q


# =============================================================================
# 13. SPECTRAL-SPATIAL CNN + QUANTUM BLOCK
# =============================================================================

class QSpectralGuard(nn.Module):

    def __init__(
        self,
        num_classes=10,
        quantum_dim=8
    ):

        super().__init__()

        # ---------------------------------------------------------------------
        # Spectral-spatial encoder
        # ---------------------------------------------------------------------

        self.encoder = nn.Sequential(

            nn.Conv2d(
                NUM_BANDS,
                32,
                kernel_size=3,
                padding=1,
                bias=False
            ),

            nn.BatchNorm2d(32),
            nn.GELU(),

            nn.Conv2d(
                32,
                64,
                kernel_size=3,
                padding=1,
                bias=False
            ),

            nn.BatchNorm2d(64),
            nn.GELU(),

            nn.MaxPool2d(2),

            nn.Conv2d(
                64,
                128,
                kernel_size=3,
                padding=1,
                bias=False
            ),

            nn.BatchNorm2d(128),
            nn.GELU(),

            nn.MaxPool2d(2),

            nn.Conv2d(
                128,
                256,
                kernel_size=3,
                padding=1,
                bias=False
            ),

            nn.BatchNorm2d(256),
            nn.GELU(),

            nn.AdaptiveAvgPool2d(1)
        )

        # ---------------------------------------------------------------------
        # Spectral bottleneck
        # ---------------------------------------------------------------------

        self.bottleneck = nn.Sequential(

            nn.Linear(
                256,
                64
            ),

            nn.GELU(),

            nn.Dropout(0.25),

            nn.Linear(
                64,
                quantum_dim
            )
        )

        # ---------------------------------------------------------------------
        # Quantum feature mapping
        # ---------------------------------------------------------------------

        self.quantum = QuantumFeatureMap(
            quantum_dim
        )

        # ---------------------------------------------------------------------
        # Final classifier
        # ---------------------------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Linear(
                quantum_dim,
                32
            ),

            nn.GELU(),

            nn.Dropout(0.20),

            nn.Linear(
                32,
                num_classes
            )
        )

    def forward(self, x):

        x = self.encoder(x)

        x = torch.flatten(
            x,
            1
        )

        x = self.bottleneck(x)

        x = self.quantum(x)

        x = self.classifier(x)

        return x


# =============================================================================
# 14. MODEL
# =============================================================================

model = QSpectralGuard(
    num_classes=NUM_CLASSES,
    quantum_dim=QUANTUM_DIM
).to(DEVICE)


# =============================================================================
# 15. MODEL INFORMATION
# =============================================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("\nModel parameters:")
print("Total:", f"{total_params:,}")
print("Trainable:", f"{trainable_params:,}")


# =============================================================================
# 16. LOSS / OPTIMIZER / SCHEDULER
# =============================================================================

criterion = nn.CrossEntropyLoss(
    label_smoothing=0.05
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=EPOCHS
)

scaler = torch.cuda.amp.GradScaler(
    enabled=torch.cuda.is_available()
)


# =============================================================================
# 17. METRIC STORAGE
# =============================================================================

history = {

    "train_loss": [],
    "val_loss": [],

    "train_acc": [],
    "val_acc": [],

    "learning_rate": [],

    "epoch_time": []
}


best_val_acc = 0.0
best_state = None


# =============================================================================
# 18. TRAINING LOOP
# =============================================================================

print("\n")
print("=" * 90)
print("TRAINING STARTED")
print("=" * 90)


for epoch in range(1, EPOCHS + 1):

    start_time = time.time()

    # -------------------------------------------------------------------------
    # TRAIN
    # -------------------------------------------------------------------------

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    train_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch:02d}/{EPOCHS} [Train]",
        leave=False
    )

    for images, targets in train_bar:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        with torch.cuda.amp.autocast(
            enabled=torch.cuda.is_available()
        ):

            outputs = model(images)

            loss = criterion(
                outputs,
                targets
            )

        scaler.scale(
            loss
        ).backward()

        scaler.step(
            optimizer
        )

        scaler.update()

        running_loss += (
            loss.item() *
            images.size(0)
        )

        predictions = outputs.argmax(
            dim=1
        )

        correct += (
            predictions == targets
        ).sum().item()

        total += images.size(0)

        train_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            acc=f"{100 * correct / total:.2f}%"
        )

    train_loss = (
        running_loss /
        total
    )

    train_acc = (
        correct /
        total
    )

    # -------------------------------------------------------------------------
    # VALIDATION
    # -------------------------------------------------------------------------

    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        val_bar = tqdm(
            val_loader,
            desc=f"Epoch {epoch:02d}/{EPOCHS} [Val]",
            leave=False
        )

        for images, targets in val_bar:

            images = images.to(
                DEVICE,
                non_blocking=True
            )

            targets = targets.to(
                DEVICE,
                non_blocking=True
            )

            with torch.cuda.amp.autocast(
                enabled=torch.cuda.is_available()
            ):

                outputs = model(images)

                loss = criterion(
                    outputs,
                    targets
                )

            val_running_loss += (
                loss.item() *
                images.size(0)
            )

            predictions = outputs.argmax(
                dim=1
            )

            val_correct += (
                predictions == targets
            ).sum().item()

            val_total += images.size(0)

    val_loss = (
        val_running_loss /
        val_total
    )

    val_acc = (
        val_correct /
        val_total
    )

    # -------------------------------------------------------------------------
    # SCHEDULER
    # -------------------------------------------------------------------------

    scheduler.step()

    lr = optimizer.param_groups[0]["lr"]

    epoch_time = time.time() - start_time

    # -------------------------------------------------------------------------
    # SAVE HISTORY
    # -------------------------------------------------------------------------

    history["train_loss"].append(
        train_loss
    )

    history["val_loss"].append(
        val_loss
    )

    history["train_acc"].append(
        train_acc
    )

    history["val_acc"].append(
        val_acc
    )

    history["learning_rate"].append(
        lr
    )

    history["epoch_time"].append(
        epoch_time
    )

    # -------------------------------------------------------------------------
    # BEST MODEL
    # -------------------------------------------------------------------------

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        best_state = copy.deepcopy(
            model.state_dict()
        )

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "val_accuracy": val_acc,
                "val_loss": val_loss,
                "class_names": class_names
            },
            CHECKPOINT_PATH
        )

        best_marker = " ★ BEST"

    else:

        best_marker = ""

    # -------------------------------------------------------------------------
    # EPOCH OUTPUT
    # -------------------------------------------------------------------------

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc*100:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc*100:.2f}% | "
        f"LR: {lr:.6f} | "
        f"Time: {epoch_time:.1f}s"
        f"{best_marker}"
    )


# =============================================================================
# 19. RESTORE BEST MODEL
# =============================================================================

if best_state is not None:

    model.load_state_dict(
        best_state
    )

print("\nBest validation accuracy:")
print(
    f"{best_val_acc*100:.2f}%"
)


# =============================================================================
# 20. SAVE TRAINING HISTORY
# =============================================================================

with open(
    os.path.join(
        OUTPUT_DIR,
        "training_history.json"
    ),
    "w"
) as f:

    json.dump(
        history,
        f,
        indent=4
    )


# =============================================================================
# 21. FINAL TEST EVALUATION
# =============================================================================

print("\n")
print("=" * 90)
print("FINAL TEST EVALUATION")
print("=" * 90)

model.eval()

all_targets = []
all_predictions = []
all_probabilities = []

test_loss = 0.0
test_total = 0

with torch.no_grad():

    for images, targets in tqdm(
        test_loader,
        desc="Testing"
    ):

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        outputs = model(images)

        loss = criterion(
            outputs,
            targets
        )

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        predictions = probabilities.argmax(
            dim=1
        )

        test_loss += (
            loss.item() *
            images.size(0)
        )

        test_total += images.size(0)

        all_targets.extend(
            targets.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.append(
            probabilities.cpu().numpy()
        )


all_targets = np.array(
    all_targets
)

all_predictions = np.array(
    all_predictions
)

all_probabilities = np.concatenate(
    all_probabilities,
    axis=0
)

test_loss /= test_total


# =============================================================================
# 22. FINAL METRICS
# =============================================================================

test_accuracy = accuracy_score(
    all_targets,
    all_predictions
)

test_precision = precision_score(
    all_targets,
    all_predictions,
    average="macro",
    zero_division=0
)

test_recall = recall_score(
    all_targets,
    all_predictions,
    average="macro",
    zero_division=0
)

test_f1 = f1_score(
    all_targets,
    all_predictions,
    average="macro",
    zero_division=0
)

weighted_precision = precision_score(
    all_targets,
    all_predictions,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    all_targets,
    all_predictions,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    all_targets,
    all_predictions,
    average="weighted",
    zero_division=0
)


print("\nFINAL TEST RESULTS")
print("-" * 60)

print(
    f"Test Loss           : {test_loss:.4f}"
)

print(
    f"Test Accuracy       : {test_accuracy*100:.2f}%"
)

print(
    f"Macro Precision     : {test_precision:.4f}"
)

print(
    f"Macro Recall        : {test_recall:.4f}"
)

print(
    f"Macro F1-score      : {test_f1:.4f}"
)

print(
    f"Weighted Precision  : {weighted_precision:.4f}"
)

print(
    f"Weighted Recall     : {weighted_recall:.4f}"
)

print(
    f"Weighted F1-score   : {weighted_f1:.4f}"
)


# =============================================================================
# 23. CLASSIFICATION REPORT
# =============================================================================

report = classification_report(
    all_targets,
    all_predictions,
    target_names=class_names,
    digits=4,
    zero_division=0
)

print("\n")
print("=" * 90)
print("CLASSIFICATION REPORT")
print("=" * 90)

print(report)

with open(
    os.path.join(
        OUTPUT_DIR,
        "classification_report.txt"
    ),
    "w"
) as f:

    f.write(report)


# =============================================================================
# 24. SAVE METRICS JSON
# =============================================================================

final_metrics = {

    "test_loss": float(test_loss),

    "test_accuracy": float(test_accuracy),

    "macro_precision": float(test_precision),

    "macro_recall": float(test_recall),

    "macro_f1": float(test_f1),

    "weighted_precision": float(weighted_precision),

    "weighted_recall": float(weighted_recall),

    "weighted_f1": float(weighted_f1),

    "best_validation_accuracy": float(
        best_val_acc
    ),

    "total_parameters": int(
        total_params
    ),

    "trainable_parameters": int(
        trainable_params
    ),

    "epochs": EPOCHS,

    "batch_size": BATCH_SIZE,

    "quantum_dimension": QUANTUM_DIM
}

with open(
    os.path.join(
        OUTPUT_DIR,
        "final_test_metrics.json"
    ),
    "w"
) as f:

    json.dump(
        final_metrics,
        f,
        indent=4
    )


# =============================================================================
# 25. PLOT 1 — TRAINING / VALIDATION LOSS
# =============================================================================

epochs_range = np.arange(
    1,
    EPOCHS + 1
)

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    epochs_range,
    history["train_loss"],
    linewidth=2,
    label="Training Loss"
)

plt.plot(
    epochs_range,
    history["val_loss"],
    linewidth=2,
    label="Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Cross-Entropy Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "01_loss_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 26. PLOT 2 — TRAINING / VALIDATION ACCURACY
# =============================================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    epochs_range,
    np.array(history["train_acc"]) * 100,
    linewidth=2,
    label="Training Accuracy"
)

plt.plot(
    epochs_range,
    np.array(history["val_acc"]) * 100,
    linewidth=2,
    label="Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "02_accuracy_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 27. PLOT 3 — LEARNING RATE
# =============================================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    epochs_range,
    history["learning_rate"],
    linewidth=2
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Learning Rate"
)

plt.title(
    "Learning Rate Schedule"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "03_learning_rate.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 28. PLOT 4 — CONFUSION MATRIX
# =============================================================================

cm = confusion_matrix(
    all_targets,
    all_predictions
)

plt.figure(
    figsize=(11, 9)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "True Class"
)

plt.title(
    "Test Set Confusion Matrix"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "04_confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 29. PER-CLASS METRICS
# =============================================================================

precision_per_class = precision_score(
    all_targets,
    all_predictions,
    average=None,
    zero_division=0
)

recall_per_class = recall_score(
    all_targets,
    all_predictions,
    average=None,
    zero_division=0
)

f1_per_class = f1_score(
    all_targets,
    all_predictions,
    average=None,
    zero_division=0
)

class_metrics = pd.DataFrame({

    "Class": class_names,

    "Precision": precision_per_class,

    "Recall": recall_per_class,

    "F1-Score": f1_per_class
})

print("\n")
print("=" * 90)
print("PER-CLASS TEST METRICS")
print("=" * 90)

print(
    class_metrics.to_string(
        index=False
    )
)

class_metrics.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "per_class_metrics.csv"
    ),
    index=False
)


# =============================================================================
# 30. PLOT 5 — PRECISION / RECALL / F1
# =============================================================================

x = np.arange(
    len(class_names)
)

width = 0.25

plt.figure(
    figsize=(14, 7)
)

plt.bar(
    x - width,
    precision_per_class,
    width,
    label="Precision"
)

plt.bar(
    x,
    recall_per_class,
    width,
    label="Recall"
)

plt.bar(
    x + width,
    f1_per_class,
    width,
    label="F1-score"
)

plt.xticks(
    x,
    class_names,
    rotation=45,
    ha="right"
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Land-Cover Class"
)

plt.title(
    "Per-Class Precision, Recall and F1-score"
)

plt.legend()

plt.ylim(
    0,
    1.05
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "05_precision_recall_f1.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 31. ROC-AUC CURVE
# =============================================================================

y_true_bin = label_binarize(
    all_targets,
    classes=np.arange(NUM_CLASSES)
)

plt.figure(
    figsize=(10, 8)
)

roc_auc_values = []

for i in range(NUM_CLASSES):

    fpr, tpr, _ = roc_curve(
        y_true_bin[:, i],
        all_probabilities[:, i]
    )

    roc_auc = auc(
        fpr,
        tpr
    )

    roc_auc_values.append(
        roc_auc
    )

    plt.plot(
        fpr,
        tpr,
        linewidth=1.5,
        label=f"{class_names[i]} (AUC={roc_auc:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    "--",
    linewidth=1
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "One-vs-Rest ROC Curves — Test Set"
)

plt.legend(
    fontsize=8,
    loc="lower right"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "06_roc_auc_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 32. MACRO ROC-AUC
# =============================================================================

macro_roc_auc = np.mean(
    roc_auc_values
)

print(
    f"\nMacro ROC-AUC: {macro_roc_auc:.4f}"
)


# =============================================================================
# 33. PRECISION-RECALL CURVES
# =============================================================================

plt.figure(
    figsize=(10, 8)
)

ap_values = []

for i in range(NUM_CLASSES):

    precision_curve, recall_curve, _ = precision_recall_curve(
        y_true_bin[:, i],
        all_probabilities[:, i]
    )

    ap = average_precision_score(
        y_true_bin[:, i],
        all_probabilities[:, i]
    )

    ap_values.append(
        ap
    )

    plt.plot(
        recall_curve,
        precision_curve,
        linewidth=1.5,
        label=f"{class_names[i]} (AP={ap:.3f})"
    )

plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "Precision-Recall Curves — Test Set"
)

plt.legend(
    fontsize=8,
    loc="lower left"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "07_precision_recall_curves.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


macro_ap = np.mean(
    ap_values
)

print(
    f"Macro Average Precision: {macro_ap:.4f}"
)


# =============================================================================
# 34. PLOT 8 — OVERALL METRICS
# =============================================================================

metric_names = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1"
]

metric_values = [
    test_accuracy,
    test_precision,
    test_recall,
    test_f1
]

plt.figure(
    figsize=(9, 6)
)

bars = plt.bar(
    metric_names,
    metric_values
)

plt.ylabel(
    "Score"
)

plt.title(
    "Overall Test Performance"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    axis="y",
    alpha=0.3
)

for bar, value in zip(
    bars,
    metric_values
):

    plt.text(
        bar.get_x() +
        bar.get_width() / 2,

        value + 0.02,

        f"{value:.4f}",

        ha="center",

        fontsize=10
    )

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "08_overall_metrics.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 35. SAMPLE TEST PREDICTIONS
# =============================================================================

model.eval()

sample_images = []
sample_labels = []
sample_predictions = []

with torch.no_grad():

    for images, targets in test_loader:

        images_device = images.to(
            DEVICE
        )

        outputs = model(
            images_device
        )

        predictions = outputs.argmax(
            dim=1
        ).cpu()

        sample_images.extend(
            images.cpu()
        )

        sample_labels.extend(
            targets
        )

        sample_predictions.extend(
            predictions
        )

        if len(sample_images) >= 12:
            break


sample_images = sample_images[:12]
sample_labels = sample_labels[:12]
sample_predictions = sample_predictions[:12]


# =============================================================================
# 36. SAMPLE PREDICTION VISUALIZATION
# =============================================================================

fig, axes = plt.subplots(
    3,
    4,
    figsize=(14, 10)
)

axes = axes.flatten()

for i in range(
    min(12, len(sample_images))
):

    image = sample_images[i]

    # RGB visualization:
    # Sentinel-2 bands approximately:
    # B4 = Red
    # B3 = Green
    # B2 = Blue

    rgb = image[
        [3, 2, 1]
    ].numpy()

    rgb = np.transpose(
        rgb,
        (1, 2, 0)
    )

    rgb = np.clip(
        rgb,
        0,
        1
    )

    axes[i].imshow(
        rgb
    )

    true_name = class_names[
        sample_labels[i].item()
    ]

    pred_name = class_names[
        sample_predictions[i].item()
    ]

    axes[i].set_title(
        f"True: {true_name}\nPred: {pred_name}",
        fontsize=9
    )

    axes[i].axis(
        "off"
    )

plt.suptitle(
    "Test Set Sample Predictions",
    fontsize=15
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "09_sample_predictions.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 37. SAVE COMPLETE RESULTS TABLE
# =============================================================================

results_summary = pd.DataFrame({

    "Metric": [
        "Test Loss",
        "Test Accuracy",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted Precision",
        "Weighted Recall",
        "Weighted F1",
        "Macro ROC-AUC",
        "Macro Average Precision"
    ],

    "Value": [
        test_loss,
        test_accuracy,
        test_precision,
        test_recall,
        test_f1,
        weighted_precision,
        weighted_recall,
        weighted_f1,
        macro_roc_auc,
        macro_ap
    ]
})

results_summary.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "final_results_summary.csv"
    ),
    index=False
)


# =============================================================================
# 38. ZIP ALL RESULTS
# =============================================================================

import shutil

zip_path = shutil.make_archive(
    "/kaggle/working/Q_SpectralGuard_Complete_Results",
    "zip",
    OUTPUT_DIR
)


# =============================================================================
# 39. FINAL SUMMARY
# =============================================================================

print("\n")
print("=" * 90)
print("Q-SPECTRALGUARD COMPLETE")
print("=" * 90)

print(
    f"Best Validation Accuracy : {best_val_acc*100:.2f}%"
)

print(
    f"Test Accuracy            : {test_accuracy*100:.2f}%"
)

print(
    f"Macro Precision          : {test_precision:.4f}"
)

print(
    f"Macro Recall             : {test_recall:.4f}"
)

print(
    f"Macro F1                 : {test_f1:.4f}"
)

print(
    f"Macro ROC-AUC            : {macro_roc_auc:.4f}"
)

print(
    f"Macro Average Precision  : {macro_ap:.4f}"
)

print(
    f"Parameters               : {total_params:,}"
)

print(
    f"Results directory        : {OUTPUT_DIR}"
)

print(
    f"ZIP file                 : {zip_path}"
)

print("\nSaved files:")

for file in sorted(
    os.listdir(OUTPUT_DIR)
):

    print(
        "  ✓",
        file
    )

print("\nExperiment finished successfully.")
```
